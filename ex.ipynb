```python
def ticket_machine():
    number = 1
    while True:  # 끝이 없어도 됨!
        yield f"{number}번 손님"
        number += 1

machine = ticket_machine()

print(next(machine))  # "1번 손님" (여기서 일시정지)
print(next(machine))  # "2번 손님" (여기서 일시정지)
print(next(machine))  # "3번 손님"
```

```python
import time

def ticket_machine():
    number = 1
    while True:
        yield f"{number}번 손님"  # 값을 내보내고 여기서 일시정지!
        number += 1
        time.sleep(1)

machine = ticket_machine()

print(next(machine))  # 1번 손님
print(next(machine))  # (1초 뒤) 2번 손님
print(next(machine))  # (1초 뒤) 3번 손님
```

```python
class TicketManager:

    def __init__(self):
        self.ticket_number: int = 1

    def ticket_machine(self) -> str:
            text = f"{self.ticket_number}번 손님"
            self.ticket_number += 1
            return text

ticket = TicketManager()

print(ticket.ticket_machine())
print(ticket.ticket_machine())
print(ticket.ticket_machine())

```

```python
korean_word:list[str] = ["가", "나", "다", "라", "마", "바", "사"]
english_word:list[str] = ["a", "b", "c", "d", "e", "f", "g"]

words:list[list] = [korean_word, english_word]

def word_machine() -> str:
    for word_list in words:
        for word in word_list:
            yield word

machine = word_machine()
print(machine)

all_word_list:list = korean_word + english_word
print(len(all_word_list))

for _ in range(len(all_word_list)):
    print(next(machine))
```

```python
import string
import secrets

strs = string.ascii_letters
print(strs)
print(type(strs))

digits = string.digits
print(digits)
print(type(digits))

random_strs = strs + digits

print(random_strs)

print(secrets.choice(random_strs))
```

```python
pre_code = []
for _ in range(6):
    choiced = secrets.choice(random_strs)
    pre_code.append(choiced)

print(pre_code)

security_code = "".join(pre_code)

print(security_code)

```
